```python
import rapids_singlecell as rsc
import scanpy as sc
import cupy as cp
import anndata as ad
```

```python
import rmm
from rmm.allocators.cupy import rmm_cupy_allocator

rmm.reinitialize(
    managed_memory=True,  # Allows oversubscription
    pool_allocator=False,  # default is False
    devices=0,  # GPU device IDs to register. By default registers only GPU 0.
)
cp.cuda.set_allocator(rmm_cupy_allocator)
```

```python
adata = sc.read("GSE250346_Seurat_GSE250346_CORRECTED_SEE_RDS_README_082024.h5ad")
del adata.obsm['X_umap']
del adata.obsm['X_pca']
del adata.varm['PCs']
```

```python
rsc.get.anndata_to_GPU(adata)
rsc.tl.pca(adata, n_comps=100)
```

```python
rsc.get.anndata_to_CPU(adata)
rsc.pp.neighbors(adata, n_neighbors=15, n_pcs=40)
rsc.tl.umap(adata)
rsc.tl.leiden(adata, resolution=0.5, key_added="leiden0.5")
```

```python
sc.pl.umap(adata, color="leiden0.5", legend_loc="on data")
```
